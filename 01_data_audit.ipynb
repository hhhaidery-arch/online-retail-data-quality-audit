```python

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# Monthly record counts
monthly_counts = (
    result["InvoiceDate"]
    .dt.to_period("M")
    .value_counts()
    .sort_index()
)

# Premium canvas
fig, ax = plt.subplots(figsize=(14, 7), facecolor="#101827")
ax.set_facecolor("#101827")

# Bar chart
bars = ax.bar(
    monthly_counts.index.astype(str),
    monthly_counts.values,
    color="#D4AF37",
    width=0.62,
    edgecolor="#F5D978",
    linewidth=0.8
)

# Titles
fig.text(
    0.09, 0.94,
    "TRANSACTION QUALITY  /  MONTHLY AUDIT",
    color="#D4AF37",
    fontsize=10,
    fontweight="bold"
)

ax.set_title(
    "Negative-Quantity Records",
    loc="left",
    color="#FFFFFF",
    fontsize=23,
    fontweight="bold",
    pad=25
)

ax.text(
    0, 1.02,
    "Zero-price, non-C invoice records • Monthly distribution",
    transform=ax.transAxes,
    color="#AAB5C5",
    fontsize=11
)

# Value labels
ax.bar_label(
    bars,
    labels=[f"{v:,}" for v in monthly_counts.values],
    padding=6,
    color="#F5F7FA",
    fontsize=10,
    fontweight="bold"
)

# Styling
ax.tick_params(axis="x", colors="#C8D0DD", labelrotation=35, labelsize=10)
ax.tick_params(axis="y", colors="#AAB5C5", labelsize=10)

ax.yaxis.grid(True, linestyle="--", alpha=0.18, color="#AAB5C5")
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

ax.set_ylabel("Number of records", color="#AAB5C5", labelpad=12)
ax.set_xlabel("Invoice month", color="#AAB5C5", labelpad=14)
ax.set_ylim(0, monthly_counts.max() * 1.20)

ax.yaxis.set_major_locator(mticker.MaxNLocator(integer=True))

fig.text(
    0.09, 0.035,
    "SOURCE: ONLINE RETAIL DATASET  |  EXPLORATORY AUDIT",
    color="#78869A",
    fontsize=9
)

plt.tight_layout(rect=[0.06, 0.07, 0.98, 0.90])
plt.show()
```

```python

import matplotlib.pyplot as plt

# Country distribution (%)
country_pct = df["Country"].value_counts(normalize=True) * 100

uk_share = country_pct.loc["United Kingdom"]
other_share = 100 - uk_share

# Premium theme
bg = "#101827"
gold = "#D4AF37"
cream = "#F5F1E8"
muted = "#AAB5C5"

fig, ax = plt.subplots(figsize=(11, 8), facecolor=bg)
ax.set_facecolor(bg)

# Donut chart: UK vs all other countries
wedges, _ = ax.pie(
    [uk_share, other_share],
    startangle=90,
    counterclock=False,
    colors=[gold, "#34445C"],
    wedgeprops=dict(width=0.28, edgecolor=bg, linewidth=3)
)

# Center text
ax.text(
    0, 0.12, f"{uk_share:.2f}%",
    ha="center", va="center",
    fontsize=29, fontweight="bold", color=cream
)
ax.text(
    0, -0.15, "UNITED KINGDOM",
    ha="center", va="center",
    fontsize=10, fontweight="bold", color=gold
)
ax.text(
    0, -0.34, "of all records",
    ha="center", va="center",
    fontsize=10, color=muted
)

# Title and subtitle
fig.text(
    0.08, 0.93, "ONLINE RETAIL  /  DATA AUDIT",
    fontsize=10, fontweight="bold", color=gold
)
fig.text(
    0.08, 0.86, "Where do transactions come from?",
    fontsize=22, fontweight="bold", color=cream
)
fig.text(
    0.08, 0.81, "Geographic concentration across the full dataset",
    fontsize=11, color=muted
)

# Legend
ax.legend(
    wedges,
    [f"United Kingdom  ·  {uk_share:.2f}%",
     f"All other countries  ·  {other_share:.2f}%"],
    loc="lower center",
    bbox_to_anchor=(0.5, -0.10),
    frameon=False,
    labelcolor=cream,
    fontsize=11
)

# Footer
fig.text(
    0.08, 0.04,
    f"TOTAL COUNTRIES: {df['Country'].nunique()}  |  SOURCE: ONLINE RETAIL DATASET",
    fontsize=9, color=muted
)

plt.tight_layout(rect=[0.04, 0.12, 0.96, 0.78])
plt.show()
```
